```python
import sentencepiece as spm

tokenizers_versions = [
    128, 256, 512, 768, 1024, 2048, 4096, 8192
]

def get_tokenizer(verb_size):
    spm_model_path = f'./tokenizers/bpe_{verb_size}.model'
    sp = spm.SentencePieceProcessor(model_file=spm_model_path)
    return sp
```

```python
tags = ['<pad>','<unk>','<s>','</s>','[PAD]','[UNK]','[CLS]','[SEP]','[MASK]']
def get_avg_length(tokenizer):
    avg_len = 0
    for i in range(tokenizer.GetPieceSize()):
        piece = tokenizer.IdToPiece(i)
        if piece in tags:
            continue
        avg_len += len(piece)
    avg_len /= tokenizer.GetPieceSize() - len(tags)
    # print('model size:', tokenizer.GetPieceSize(), ', average length:', avg_len)
    return avg_len
```

```python
# draw a table of average length of tokens
print('model size, average length')
for verb_size in tokenizers_versions:
    tokenizer = get_tokenizer(verb_size)
    avg_len = get_avg_length(tokenizer)
    print(f'{verb_size}, {avg_len}')
```

```python
hg38_path = '/home/huangyi/data_m/compress_dataset/train/hg38.fa.ob'
from tqdm import tqdm

max_seq_len = 512
def get_tokenized_length(tokenizer):
    tokenized_len = 0
    with open(hg38_path, 'r') as f:
        raw_datas = []
        lines = f.readlines()
        for line in tqdm(lines, desc=f'prehandle file hg38'):
            # line = line.decode('utf-8').strip()
            raw_datas.append(line)
        raw_data = ''.join(raw_datas)
        raw_len = len(raw_data)
        raw_datas = None
        tokens = []
        print('raw length of hg38:', raw_len)

        for i in tqdm(range(0, len(raw_data), max_seq_len), desc=f'tokenizing file hg38'):
            token_seq = tokenizer.EncodeAsIds(raw_data[i: i+max_seq_len])
            tokens += token_seq

        tokenized_len = len(tokens)
        print('tokenized length of hg38:', tokenized_len)
    return raw_len / tokenized_len
```

```python
tokenized_lens = []
for verb_size in tokenizers_versions:
    tokenizer = get_tokenizer(verb_size)
    print(f'processing model size {verb_size}')
    avg_len = get_tokenized_length(tokenizer)
    tokenized_lens.append(avg_len)

print('model size, tokenized length')
for i in range(len(tokenizers_versions)):
    print(f'{tokenizers_versions[i]}, {tokenized_lens[i]}')
```
